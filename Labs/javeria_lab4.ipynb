```python
def fuuast_agent(user_input):

    course = {
        "OOP": {
            "batch": "22, 23, 24",
            "course": "OOP",
            "teacher": "Ms. Noshaba"
        },

        "Database": {
            "batch": "24, 25, 26",
            "course": "Database",
            "teacher": "DR. Uzma"
        },

        "Networking": {
            "batch": "25, 26, 27",
            "course": "Networking",
            "teacher": "Ms. Asima"
        }
    }

    prompt = f"""
You are a professional guuide agent for Federal Urdu University.

If user is asking about course info ask user which course and look from dictionary students and course and reply accordingly.
If user input is generaric reply otherwise
if information user is asking not generic or not in dictionary
reply: I don't have information regarding this.

Notify all the course detail related to user's input:

1. Batch Information
2. Course Information
3. Course Handler Information


Make response brief and easy for user.

User request: {user_input}

Course Guide:
{course}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST Course Guide Agent")
print("Type 'exit' to stop\n")

while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = fuuast_agent(user)
        print("\nAgent:", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
